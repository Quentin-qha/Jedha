```python
# precent of missing values
df_missing_values = df.isna().sum() / len(df) * 100
df_missing_values_sort = df_missing_values[df_missing_values >= 50].sort_values(ascending=False).reset_index()
df_missing_values_sort.columns = ['col_name', 'percent_missing']
df_missing_values_sort['percent_missing'] = df_missing_values_sort['percent_missing'].round(2)
df_missing_values_sort
```

```python
fig = px.scatter(df_missing_values.values)
fig.update_layout(
    title='Percentage of Missing Values in Each Column (by index)',
    xaxis_title='Row Index',
    yaxis_title='% Missing Value',
    showlegend=False,
)
fig.show()
```

```python
def color_code(percent):
    if percent > 70:
        return '#FB2C36'
    elif percent > 50:
        return '#FE9A00'
    else:
        return '#7CCF00'

df_missing_values_sort['color'] = df_missing_values_sort['percent_missing'].apply(color_code)

fig = px.bar(
    df_missing_values_sort,
    x='col_name',
    y='percent_missing',
    color='color',
    color_discrete_map='identity',
    title='Percentage of missing values in each column (by name)',
)

fig.update_layout(
    xaxis_title='Column',
    yaxis_title='Percentage Missing',
    xaxis_tickangle=-45
)

fig.show()
```

```python
# All columns name and type
df_type = df.dtypes.reset_index().rename(columns={"index": "c_name", 0: "type"})
pd.set_option('display.max_rows', 200)
df_type
```

```python
# renaming my column name for better understanding
renames = {
    'condtn'    : 'nbr_of_partner',
    'field_cd' : 'field_c' #because carreer code is career_c 
}

# Apply the renames
df.rename(columns=renames, inplace=True)
```

```python
# Fill data if I can 
# Apply a wave id to the only one missing NaN => df[df['iid'] == 552]
df['id'] = df['id'].fillna(22)

# num_in_3 92% missing values but the values is "if yes, how many?" than if no value that will say 0
df['num_in_3'] = df['num_in_3'].fillna(0)
```

```python
# Columns to convert to Int64
col_to_convert_to_int = [
    'positin1', 
    'pid', 
    'age_o', 
    'age'
]

# Convertion apply to be an int
df['id'] = df['id'].astype(int)
df['num_in_3'] = df['num_in_3'].astype(int)
df[col_to_convert_to_int] = df[col_to_convert_to_int].astype('Int64')
```

```python
# Mapping dictionnaires
gender_mapping = {0: 'Female', 1 : 'Male'}

wave_date_mapping = {
    1: '2002-10-16',
    2: '2002-10-23',
    3: '2002-11-12',
    4: '2002-11-12',
    5: '2002-11-20',
    6: '2003-03-26',
    7: '2003-03-26',
    8: '2003-04-02',
    9: '2003-04-02',
    10: '2003-09-24'
}

race_mapping = {
    1 : 'Black/African American',
    2 : 'European/Caucasian-American',
    3 : 'Latino/Hispanic American',
    4 : 'Asian/Pacific Islander/Asian-American',
    5 : 'Native American',
    6 : 'Other'
}

frequency_maping = {
    1 : 'Several times a week',
    2 : 'Twice a week',
    3 : 'Once a week',
    4 : 'Twice a month',
    5 : 'Once a month',
    6 : 'Several times a year',
    7 : 'Almost never'
}

# Apply mapping and category type
df['gender'] = df['gender'].map(gender_mapping).astype('category')
df['race_o'] = df['race_o'].map(race_mapping).astype('category')
df['race'] = df['race'].map(race_mapping).astype('category')
df['go_out'] = df['go_out'].map(frequency_maping).astype('category')
df['date'] = df['date'].map(frequency_maping).astype('category')
df['wave_date'] = df['wave'].map(wave_date_mapping)

# Apply dateTime to date column
df['wave_date'] = pd.to_datetime(df['wave_date'])

# Apply category type
df['field'] = df['field'].astype('category')
```

```python
# Drop columns with more than 50% missing values
df = df.drop(df_missing_values[df_missing_values >= 50].index, axis=1)
df.shape
```

### 4.4 Predicting one's own perceived value
**Question:**\
Can people accurately predict their own perceived value in the dating market?

**Approach:** 
* Comparison between self-evaluation (`attr3_1`, `sinc3_1`, etc.) and evaluation by others (`attr`, `sinc`, etc.).
* Calculation of the average percentage difference.

**Result:** \
Participants **overestimate_** their attractiveness by about **_8 to 13% on average_**.

```python
df_attributes_me_vs_other = df[['attr', 'sinc', 'fun', 'intel', 'amb', 'attr3_1', 'sinc3_1', 'fun3_1', 'intel3_1', 'amb3_1']].copy()
df_attributes_me_vs_other.sample(10)
```

```python
array = []
df_attributes_me_vs_other['attr_mean'] = df_attributes_me_vs_other['attr'] - df_attributes_me_vs_other['attr3_1']
df_attributes_me_vs_other['sinc_mean'] = df_attributes_me_vs_other['sinc'] - df_attributes_me_vs_other['sinc3_1']
df_attributes_me_vs_other['fun_mean'] = df_attributes_me_vs_other['fun'] - df_attributes_me_vs_other['fun3_1']
df_attributes_me_vs_other['intel_mean'] = df_attributes_me_vs_other['intel'] - df_attributes_me_vs_other['intel3_1']
df_attributes_me_vs_other['amb_mean'] = df_attributes_me_vs_other['amb'] - df_attributes_me_vs_other['amb3_1']

print('Attractiveness difference:', abs(round(df_attributes_me_vs_other['attr_mean'].mean() * 10, 2)), '%')
print('Sincerity difference:', abs(round(df_attributes_me_vs_other['sinc_mean'].mean() * 10, 2)), '%')
print('Fun difference:', abs(round(df_attributes_me_vs_other['fun_mean'].mean() * 10, 2)), '%')
print('Intelligence difference:', abs(round(df_attributes_me_vs_other['intel_mean'].mean() * 10, 2)), '%')
print('Ambition difference:', abs(round(df_attributes_me_vs_other['amb_mean'].mean() * 10, 2)), '%')
```

```python
attributes_peers = [
    ('attr', 'attr3_1'),
    ('sinc', 'sinc3_1'),
    ('intel', 'intel3_1'),
    ('fun', 'fun3_1'),
    ('amb', 'amb3_1')
]

fig = make_subplots(rows=1, cols=5, subplot_titles=[f"{a}" for a, _ in attributes_peers])

for i, (col1, col2) in enumerate(attributes_peers):
    fig.add_trace(
        go.Bar(
            name='Others',
            x=['Others'],
            y=[df[col1].mean()],
            marker_color='steelblue',
            showlegend= True if i == 0 else False
        ),
        row=1, col=i+1
    )
    fig.add_trace(
        go.Bar(
            name='Self-rated',
            x=['Self'],
            y=[df[col2].mean()],
            marker_color='orange',
            showlegend= True if i == 0 else False
        ),
        row=1, col=i+1
    )

# Layout
fig.update_layout(
    height=400,
    width=1200,
    title_text='Self vs Perceived Rating Comparison (Mean)',
    barmode='group'
)

fig.show()
```

![Chart as image](../speed_dating__tinder/charts_as_png/4_4_self_vs_perceived_rating_comparison.png)

### 4.5 Impact of running order (first or last)
**Question:** \
In terms of getting a second date, is it better to be someone's first speed date of the night or their last?

**Approach:** 
* Use the order variable to identify the first and last speed date.
* Comparison of second date rates by order.

**Results:** \
Little noticeable impact found: order does not seem to significantly influence decisions.

```python
df_first_date = df[df['order'] == 1]
df_pie_first_date = df_first_date['match'].value_counts().reset_index()
df_pie_first_date['match_label'] = df_pie_first_date['match'].map({0: 'No match', 1: 'Match'})


df_last_date = df[df['order'] == df.groupby('iid')['order'].transform('max')]
df_pie_last_date = df_last_date['match'].value_counts().reset_index()
df_pie_last_date['match_label'] = df_pie_last_date['match'].map({0: "No match", 1: "Match"})

fig = make_subplots(rows=1, cols=2, specs=[[{'type':'domain'}, {'type':'domain'}]], subplot_titles=['First date', 'Last date'])

fig.add_trace(
    go.Pie(
        labels=df_pie_first_date['match_label'],
        values=df_pie_first_date['count'],
        name="First date"
    ),
    row=1, col=1
)

fig.add_trace(
    go.Pie(
        labels=df_pie_last_date['match_label'],
        values=df_pie_last_date['count'],
        name="Last date"
    ),
    row=1, col=2
)

fig.update_layout(
    title_text="First date vs Last date - match conversion"
)

fig.show()
```
