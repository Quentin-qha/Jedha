2bis. Produce a list giving the indices of all couples of variables that are correlated above 90%

```python
corr_matrix = data.corr().abs()

np.fill_diagonal(corr_matrix.values, 0)

high_corr_pairs = [
    (col1, col2)
    for col1 in corr_matrix.columns
    for col2 in corr_matrix.columns
    if corr_matrix.loc[col1, col2] > 0.9
]

high_corr_pairs
```

```python
unique_pairs = set(tuple(sorted(pair)) for pair in high_corr_pairs)
unique_pairs = list(unique_pairs)
unique_pairs
print(f"Colonnes to bes droped: '{unique_pairs[0][0]}, {unique_pairs[1][0]}, {unique_pairs[2][0]}'")
```
