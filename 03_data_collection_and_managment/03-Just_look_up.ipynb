```python
# Calculate the date of the last 24 hours
today = datetime.now()
now_str = today.strftime("%Y-%m-%dT%H:%M:%SZ")
yesterday = today - timedelta(days=1)
yesterday_str = yesterday.strftime("%Y-%m-%dT%H:%M:%SZ")
```

```python
location_id = data_first_location['results'][0]['id']
params = {
    "datetime_from": yesterday_str,
    "datetime_to": now_str,
    "location_id": location_id,
    "limit": 1000
}


r_Last_24h = requests.get(f"https://api.openaq.org/v3/locations/{location_id}", headers=headers, params=params)
data_Last_24h = r_Last_24h.json()
data_Last_24h['results']
```

11. As you may see, a query covering 10 days for a single location already contains 46920 measures in this case. Imagine if we would like to get historical data for the prrevious ten years for the whole country, this would take some time and a lot of space, so for now just try and create a loop to get 30 days of data for the whole country.

```python
base_url = "https://api.openaq.org/v3/locations"
location_id = data_first_location['results'][0]['id']

all_results = []
for i in range(30):
    date_to = datetime.now() - timedelta(days=i)
    date_from = date_to - timedelta(days=1)

    params = {
        "location_id": location_id,
        "datetime_from": date_from.strftime("%Y-%m-%dT%H:%M:%SZ"),
        "datetime_to": date_to.strftime("%Y-%m-%dT%H:%M:%SZ"),
        "limit": 1000
    }

    response = requests.get(base_url, headers=headers, params=params)
    data = response.json()
    all_results.extend(data.get("results", []))
```
