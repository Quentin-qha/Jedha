6. Based on the results obtained, create a list called `url_list` containing all the urls for the movies in the charts.

```python
import pandas as pd
import json

with open('01-Become_a_movie_director/imdb1.json', 'r') as f:
    json_result = json.load(f)

json_result
```

```python
df = pd.DataFrame(json_result)
df["nb_voters"] = df["nb_voters"].apply(lambda x: f"${x[1]}" if isinstance(x, list) and len(x) > 1 else None)

df
```

```python
url_list = []

for i in range(len(df)):
    url = df.loc[i]['url']
    url_list.append(url)
url_list
```

7. Store the list of urls to a file called `url_list.txt`

```python
with open("01-Become_a_movie_director/url_list.txt", "w") as f:
    for url in url_list:
        f.write(url + "\n")
```
